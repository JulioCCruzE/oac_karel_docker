# Sistema de evaluación

Hata este momento solo nos hemos centrado en karel, pero ya habrán notado que el simulador carga un circuito con algunos obstaculos. 

El proposito de este circuito es poner a prueba la navegación autónoma de karel. A grandes rasgos, la prueba consiste en que karel debe navegar por el circuito desde su posición inicial hasta la meta y en el proceso evitar chocar con los obstaculos.

Para medir el desempeño del robot al ralizar esta prueba se creó un sistema de evaluación que se encuentra en la sección **Evaluacion**. Aquí puedes encontrar las siguientes funciones: 

- El botón `Iniciar` inicia el cronométro, la detección de obstaculos, si aumenta o disminuye el puntaje, y si se alcanzó la meta.
- El botón `Reiniciar evaluación` deja el puntaje y el cronómetro en cero, además se dejara de contabilizar el puntaje.
- El botón `Volver al inicio` regresa al robot a su posición inicial. Esta función no reiniciar la evaluación.
- El input box `Límite` sirve para indicar el tiempo maximo para completar la prueba y una cuenta hacía atrás indica el tiempo restante.
- El checkbox `Sin límite de tiempo`  sirve para habilitar o deshabilitar el límite de tiempo para realizar la prueba.
- Se muestran algunos mensajes informativos, por ejemlo "Medición iniiada" o "Evaualción reiniciada".
- Un indicador que muestra cuanto tiempo ha durado la prueba. Se detiene automáticamente al llegar a la meta.
- Un indicador de puntaje.

## Puntaje

1. El puntaje aumeta al evadir obstáculos y llegar a la meta. 



2. El puntaje solo disminuye al colisionar con un obstáculo.
Si colisiona n veces con el mismo obstaculo, el puntaje se reducirá n veces.



<div class="alert alert-block alert-warning">
