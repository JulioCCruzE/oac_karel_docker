Para iniciar la ejecución del emulador del procesador se debe ocupar el IDE y una terminal de VS Code:

2. Recuerda que el IDE con VS Code se encuentra en http://localhost:3001

## Movimiento líneal

Para controlar el movimiento líneal se ocupa el registro `[C000] vel_linear`:

- Usa valores positivos para que el robot avance hacia adelante. 
- Por ejemplo, con el valor `0AH` el robot avanzará hacía enfrente con una velocidad de $10 cm/s$
- Usa valores negaticos para que el robot avance hacia atrás. 
- Por ejemplo, con el valor `F6H` el robot avanzará hacía atrás con una velocidad de $10 cm/s$

## Movimiento angular

Para controlar el movimiento líneal se ocupa el registro `[C001] vel_angular`:

- Usa valores positivos para que el robot gire a la izquierda. 
- Por ejemplo, con el valor `0AH` el robot girará a la izquierda con una velocidad de $10 grados/s$
- Usa valores negaticos para que el robot gire a la derecha. 
- Por ejemplo, con el valor `F6H` el robot girará a la derecha con una velocidad de $10 grados/s$

## Escribir en memoria RAM

Puedes realizar pruebas de movimiento modificando directamente los registros en memoria RAM:



El movimiento líneal tiene **prioridad** sobre el movimiento angular, es decir, para realizar los giros la velocidad líneal debe ser **cero**.



<div class="alert alert-block alert-warning">
